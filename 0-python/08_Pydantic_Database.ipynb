```python
from pydantic import BaseModel, Field, ValidationError

class Customer(BaseModel):
    customer_id: int
    name: str = Field(min_length=1)
    email: str

valid = Customer.model_validate({'customer_id': '1001', 'name': 'Ravi', 'email': 'ravi@example.com'})
print(valid)

try:
    Customer.model_validate({'customer_id': 'abc', 'name': '', 'email': 'bad'})
except ValidationError as error:
    print(error)
```

```python
from typing import Literal

class Order(BaseModel):
    order_id: int
    customer_id: int
    status: Literal['Pending', 'Shipped', 'Delivered']

order = Order.model_validate({'order_id': 5001, 'customer_id': 1001, 'status': 'Shipped'})
print(order.model_dump())
```
