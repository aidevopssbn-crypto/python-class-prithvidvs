# Day 01 — Variables, Types & f-strings

This notebook introduces Python fundamentals used in LLM engineering.

**Topics:** variables, int, float, str, bool, None, f-strings, lists, dicts


```python
# Variables and basic types
customer_name = "Danielle"
order_id = "ORD-3042"
max_tokens = 1024
temperature = 0.2
is_verified = True

print(type(customer_name))
print(type(max_tokens))
print(type(temperature))
print(type(is_verified))
```

```python
# f-strings: a core pattern for building prompts
prompt = f"Customer: {customer_name}\nOrder: {order_id}\nTemperature: {temperature}\nAnswer in 2 sentences."
print(prompt)
```
