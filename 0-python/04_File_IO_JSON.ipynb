**Goal:** Learn to read and write files, understand JSON, and process simple CSV data.


```python
import csv
import json

config = {'model': 'gpt-4o', 'max_tokens': 1024, 'temperature': 0.2}
raw = '{"category": "TRACK_ORDER", "confidence": "high"}'
parsed = json.loads(raw)
print(parsed)

with open('model_config.json', 'w', encoding='utf-8') as f:
    json.dump(config, f, indent=2)

with open('model_config.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)

print(loaded)
```

```python
with open('sample_prompt.txt', 'w', encoding='utf-8') as f:
    f.write('## Role\nYou are a customer support assistant.\n')

with open('sample_prompt.txt', 'r', encoding='utf-8') as f:
    prompt = f.read()
    print(prompt)
```

```python
with open('customers.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['customer_id', 'name', 'city'])
    writer.writerow([1001, 'Ravi', 'Hyderabad'])
    writer.writerow([1002, 'Meera', 'Bengaluru'])

with open('customers.csv', 'r', newline='', encoding='utf-8') as f:
    rows = csv.DictReader(f)
    for row in rows:
        print(row['name'], row['city'])
```
