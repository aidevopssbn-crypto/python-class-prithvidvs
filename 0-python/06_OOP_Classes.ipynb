## Learning goals
- Explain a class, an object, `__init__`, and `self`
- Store data and behavior together
- Write instance methods and use `@property`


```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def show_details(self):
        print(f"{self.name} works in {self.department} and earns ₹{self.salary:,}")

employee = Employee('Ananya', 75000, 'Data')
employee.show_details()
```

```python
class Customer:
    def __init__(self, name, email):
        self.name = name
        self.email = email
        self.total_spend = 0

    def place_order(self, amount):
        self.total_spend += amount

    def show_summary(self):
        print(f"Customer: {self.name}, Total spend: ₹{self.total_spend:,}")

customer = Customer('Meera', 'meera@example.com')
customer.place_order(1200)
customer.show_summary()
```
