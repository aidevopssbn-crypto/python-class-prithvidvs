# Day 03 — Functions and Data Structures

Python functions and collections are essential for agent workflows and data transformation.


```python
def build_prompt(customer_name: str, order_id: str) -> str:
    return f"Customer {customer_name} has order {order_id}. Summarize the status."

print(build_prompt("Prudhvi", "ORD-1001"))
```

```python
products = [
    {"id": 1, "name": "Laptop", "price": 1200},
    {"id": 2, "name": "Mouse", "price": 49},
]

print(products[0]["name"])
print(products[1]["price"])
```
