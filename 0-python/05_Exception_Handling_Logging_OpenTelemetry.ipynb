```python
import json
import logging
import time

logging.basicConfig(level=logging.INFO, format='%(levelname)s:%(message)s')

def safe_to_int(value):
    try:
        return int(value)
    except ValueError:
        logging.warning('Could not convert %r to int', value)
        return None

print(safe_to_int('1001'))
print(safe_to_int('hello'))
```

```python
def validate_customer_id(customer_id):
    try:
        value = int(customer_id)
    except (TypeError, ValueError) as error:
        raise ValueError('Customer ID must be a whole number.') from error
    if value <= 0:
        raise ValueError('Customer ID must be greater than zero.')
    return value

for test in ['1001', 'abc', '-5']:
    try:
        print(validate_customer_id(test))
    except ValueError as error:
        print('Invalid ID:', error)
```
