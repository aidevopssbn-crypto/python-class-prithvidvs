# Version 1 — Prompt Anatomy: Brain, No Hands

This notebook introduces a simple AI data analyst agent that can generate SQL but has no tools.


```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
import os
from dotenv import load_dotenv

load_dotenv()

if not os.environ.get("OPENAI_API_KEY"):
    raise ValueError("Set OPENAI_API_KEY in your environment or .env file")

model = ChatOpenAI(model="gpt-4o-mini")
agent = create_agent(model=model, tools=[], system_prompt="You are an AI data analyst. Return SQL only if the schema is known.")

result = agent.invoke({"messages": [{"role": "user", "content": "How many orders were cancelled last month?"}]})
print(result["messages"][-1].content)
```
