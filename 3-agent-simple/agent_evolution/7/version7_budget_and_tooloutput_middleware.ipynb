# Version 7 — Tool Output Middleware

This notebook trims large tool outputs and keeps answer generation under control.


```python
def summarize_tool_output(output: str, max_chars: int = 3000):
    if len(output) <= max_chars:
        return output
    return output[:max_chars] + '
...[truncated]'

large = 'x' * 10000
print(len(summarize_tool_output(large)))
```
