# Version 6 — Budget Middleware

This notebook adds a middleware budget constraint so the agent can spend only a limited number of tool calls per question.


```python
class BudgetMiddleware:
    def __init__(self, max_tool_calls=3):
        self.max_tool_calls = max_tool_calls
        self.used = 0

    def check(self):
        if self.used >= self.max_tool_calls:
            raise RuntimeError('Budget exhausted: no more tool calls allowed.')
        self.used += 1
        return True

budget = BudgetMiddleware()
print(budget.check())
print(budget.check())
```
