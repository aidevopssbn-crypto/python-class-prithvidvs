# Version 4 — Query Validation + Workflow

This notebook adds a validation step so the agent checks queries before executing them.


```python
from langchain_core.tools import tool

@tool
def validate_query(sql: str) -> str:
    sql_clean = sql.strip()
    if sql_clean.lower().startswith('select'):
        return 'VALID: read-only SELECT statement looks acceptable.'
    return 'INVALID: only SELECT queries are allowed for this workflow.'

print(validate_query('SELECT * FROM orders'))
```
